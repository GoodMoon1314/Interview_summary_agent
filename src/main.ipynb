```python
from langgraph.checkpoint.postgres import PostgresSaver
from agent.graph.graph import builder
from config.config import db_config

config = {
    "configurable":{
        "thread_id":"test_9_18_3"
    }
}

with PostgresSaver.from_conn_string(db_config["DB_URL"]) as checkpointer:
    main_graph = builder.compile(checkpointer=checkpointer)


    res = main_graph.invoke(
        {
            "user_input": "agent开发",
            "is_history": False
        },
        config=config
    )


    print(f"输出结果:{type(res)}")
    for i in res:
        print(i)


    from IPython.display import display,Image

    display(
        Image(main_graph.get_graph(xray=True).draw_mermaid_png())
    )
```
